```python
import pandas as pd
df = pd.read_csv('../datasets/non-eu/Orders.csv', sep=';')
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
```

```python
df.info()
```

```python
cutoff_date = df['Order Date'].max() - pd.Timedelta(days=7)
print(cutoff_date)
```

```python
last_week_sales = df.loc[df['Order Date'] > cutoff_date, ['Sales']].sum()
last_week_sales_avg = df.loc[df['Order Date'] > cutoff_date, ['Sales']].mean()
print(last_week_sales)
print(last_week_sales_avg)
```
